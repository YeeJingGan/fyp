```python
import nibabel as nib
import matplotlib.pyplot as plt
import numpy as np
```

```python
t1_path = r"C:\Users\hp\Downloads\BraTs-T1\BraTS20_Training_369_t1.nii.gz"
t2_path = r"C:\Users\hp\Downloads\BraTs-T2\BraTS20_Training_369_t2.nii.gz"
gt_path = r"C:\Users\hp\Downloads\BraTs-GT\BraTS20_Training_369_seg.nii.gz"
```

```python
# Paths to the .nii.gz files
file_paths = [
    t1_path,
    t2_path,
    gt_path
]

# Initialize subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Loop through the files and visualize each
for i, file_path in enumerate(file_paths):
    # Load the .nii.gz file
    img = nib.load(file_path)
    
    # Get the image data as a numpy array
    data = img.get_fdata()
    
    # Select a slice (e.g., the middle slice along the first axis)
    slice_index = data.shape[0] // 2
    slice_data = data[slice_index, :, :]
    
    # Plot the slice on the corresponding subplot
    axes[i].imshow(slice_data.T, cmap="gray", origin="lower")
    axes[i].set_title(f"Slice {slice_index} - File {i+1}")
    axes[i].axis('off')  # Turn off axes for cleaner display

# Show the plots
plt.tight_layout()
plt.show()
```
