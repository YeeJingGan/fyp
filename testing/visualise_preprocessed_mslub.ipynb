```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
import torch
```

```python
mri1_path = r"C:\Users\hp\OneDrive\Documents\fyp\data\MSLUB-T1\patient16_T1W.npz"
mri2_path = r"C:\Users\hp\OneDrive\Documents\fyp\data\MSLUB-T2\patient16_T2W.npz"

data_1 = np.load(mri1_path)  # Load .npz file
data_2 = np.load(mri2_path)

img1 = torch.tensor(data_1["image"]).squeeze().numpy()
img2 = torch.tensor(data_2["image"]).squeeze().numpy()

# Display the image and anomaly slices
fig, axes = plt.subplots(1, 3, figsize=(10, 5))
image_slice = img1[:, :, 62]
anomaly_slice = img2[:, :, 62]
            
axes[0].imshow(image_slice, cmap="gray")
axes[1].imshow(anomaly_slice, cmap="gray")
axes[2].imshow(image_slice, cmap="gray")
axes[2].imshow(anomaly_slice, cmap="hot", alpha=0.5)

 
for col in range(3):
    axes[col].axis("off")

            
axes[0].set_title("MRI")
axes[1].set_title("Anomaly")
axes[2].set_title("MRI + Anomaly")

plt.tight_layout()
plt.show()
```
