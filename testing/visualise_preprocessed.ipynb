```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
import torch
```

```python
mri1_path = r"C:\Users\hp\OneDrive\Documents\fyp\data\IXI-T1\IXI002-Guys-0828-T1.npz"
mri2_path = r"C:\Users\hp\OneDrive\Documents\fyp\data\IXI-T2\IXI002-Guys-0828-T2.npz"

data_1 = np.load(mri1_path)  # Load .npz file
data_2 = np.load(mri2_path)

img1 = torch.tensor(data_1["image"]).squeeze(0).numpy()
img2 = torch.tensor(data_2["image"]).squeeze(0).numpy()

noisy_image1 = torch.tensor(data_1["noisy_image"]).squeeze(0).numpy()
noisy_image2 = torch.tensor(data_2["noisy_image"]).squeeze(0).numpy()

fig = plt.figure(figsize=(10, 10))
gs = gridspec.GridSpec(2, 2)

ax1 = fig.add_subplot(gs[0, 0])
ax1.imshow(img1[64,:,:], cmap='gray')
ax1.set_title('T1 Original')

ax2 = fig.add_subplot(gs[0, 1])
ax2.imshow(noisy_image1[64,:,:], cmap='gray')
ax2.set_title('T1 Noisy')

ax3 = fig.add_subplot(gs[1, 0])
ax3.imshow(img2[64,:,:], cmap='gray')
ax3.set_title('T2 Original')

ax4 = fig.add_subplot(gs[1, 1])
ax4.imshow(noisy_image2[64,:,:], cmap='gray')
ax4.set_title('T2 Noisy')

plt.show()
```
