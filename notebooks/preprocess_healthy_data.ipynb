**Import Libraries**

```python
import os
import dotenv
dotenv.load_dotenv()
import sys
sys.path.append(os.getenv('PROJECT_ROOT'))
import numpy as np
from src.data_preprocessor import DataPreprocessor
```

**Preprocessing**

```python
# Define raw data path
t1_path = os.path.join(os.getenv('RAW_DATA_DIR'), 'IXI-T1')
t2_path = os.path.join(os.getenv('RAW_DATA_DIR'), 'IXI-T2')

# Define preprocessed data path
preprocessed_t1_path = os.path.join(os.getenv('PROJECT_ROOT'), 'data\IXI-T1')
preprocessed_t2_path = os.path.join(os.getenv('PROJECT_ROOT'), 'data\IXI-T2')
```

```python
# Preprocess T1 data
t1_count = 0

for t1_file in os.listdir(t1_path):
    file = os.path.join(t1_path, t1_file)
    img_without_noise, img_with_noise = DataPreprocessor().preprocess(mri_path=file, mdl=1, add_noise=True)
    save_path = os.path.join(preprocessed_t1_path, t1_file.replace('.nii.gz', '.npz'))
    np.savez_compressed(save_path, 
                        image=img_without_noise,
                        noisy_image=img_with_noise)
    t1_count += 1

if(t1_count == len(os.listdir(t1_path))):
    print('T1 data preprocessing completed')
else:
    print('There are some files missing in T1 data preprocessing')
```

```python
# Preprocess T2 data
t2_count = 0

for t2_file in os.listdir(t2_path):
    file = os.path.join(t2_path, t2_file)
    img_without_noise, img_with_noise = DataPreprocessor().preprocess(mri_path=file, mdl=2, add_noise=True)
    save_path = os.path.join(preprocessed_t2_path, t2_file.replace('.nii.gz', '.npz'))
    np.savez_compressed(save_path, 
                        image=img_without_noise,
                        noisy_image=img_with_noise)
    t2_count += 1

if(t2_count == len(os.listdir(t2_path))):
    print('T2 data preprocessing completed')
else:
    print('There are some files missing in T2 data preprocessing')
```
